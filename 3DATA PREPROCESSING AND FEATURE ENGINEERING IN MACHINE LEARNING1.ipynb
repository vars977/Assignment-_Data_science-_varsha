```python
import streamlit as st
import pandas as pd 
df=pd.read_csv(r'C:\Users\lenovo\Desktop\Assigment\adult_with_headers (1).csv',header=0)
```

```python
df.info()
```

```python
st.subheader("Original Dataset")
st.dataframe(df.head())
```

```python
st.subheader("Missing Values")
st.write(df.isnull().sum())
```

```python
df.ffill(inplace=True)
```

```python
# Numerical Columns
num_cols = df.select_dtypes(include=["int64","float64"]).columns

# Categorical Columns
cat_cols = df.select_dtypes(include=["object"]).columns
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
option = st.selectbox(
    "Choose Scaling",
    ["None","Standard Scaling","Min-Max Scaling"]
)

if option == "Standard Scaling":
    scaler = StandardScaler()
    df[num_cols] = scaler.fit_transform(df[num_cols])

elif option == "Min-Max Scaling":
    scaler = MinMaxScaler()
    df[num_cols] = scaler.fit_transform(df[num_cols])

encoding = st.selectbox(
    "Choose Encoding",
    ["None","Label Encoding","One-Hot Encoding"]
)

if encoding == "Label Encoding":
    le = LabelEncoder()

    for col in cat_cols:
        df[col] = le.fit_transform(df[col].astype(str))

elif encoding == "One-Hot Encoding":
    df = pd.get_dummies(df, columns=cat_cols, dtype=int)

st.subheader("Processed Dataset")
st.dataframe(df.head())

st.download_button(
    "Download Processed Data",
    df.to_csv(index=False),
    file_name="processed_data.csv",
    mime="text/csv"
)
```
